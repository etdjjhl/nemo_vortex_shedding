# 01 — Dataset Exploration

Inspect the cylinder_flow dataset structure, visualize sample meshes and ground-truth flow fields.

```python
import sys
sys.path.insert(0, '..')  # project root

import numpy as np
import matplotlib.pyplot as plt
from matplotlib import tri as mtri

from physicsnemo.datapipes.gnn.vortex_shedding_dataset import VortexSheddingDataset
```

```python
DATA_DIR = '../raw_dataset/cylinder_flow/cylinder_flow'

dataset = VortexSheddingDataset(
    name='explore',
    data_dir=DATA_DIR,
    split='train',
    num_samples=5,
    num_steps=10,
)
print(f'Samples: {len(dataset)}')
```

```python
# Inspect a single graph sample
graph, cells, mask = dataset[0]
print('Node features (x) shape:', graph.x.shape)
print('Edge features shape    :', graph.edge_attr.shape)
print('Target (y) shape       :', graph.y.shape)
print('Mesh positions shape   :', graph['mesh_pos'].shape)
print('Cells shape            :', cells.shape)
```

```python
# Visualize the mesh for the first sample
pos = graph['mesh_pos'].numpy()
triang = mtri.Triangulation(pos[:, 0], pos[:, 1], cells.squeeze().numpy())

fig, ax = plt.subplots(figsize=(12, 4))
ax.triplot(triang, 'b-', lw=0.3, alpha=0.6)
ax.set_title('Sample mesh (cylinder_flow)')
ax.set_aspect('equal')
plt.tight_layout()
plt.show()
print(f'Number of nodes: {pos.shape[0]},  Number of triangles: {cells.shape[-1]}')
```

```python
# Visualize ground-truth u velocity at t=0
# Node features: [vx, vy, node_type, x, y, ?] — first two cols are velocity
stats = dataset.node_stats
u = graph.x[:, 0].numpy()  # normalised

fig, ax = plt.subplots(figsize=(12, 4))
tc = ax.tripcolor(triang, u, cmap='RdBu_r')
plt.colorbar(tc, ax=ax, label='u (normalised)')
ax.set_title('Ground truth u-velocity at t=0')
ax.set_aspect('equal')
plt.tight_layout()
plt.show()
```
